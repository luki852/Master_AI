# PyMOL
PyMOL is a tool that is used to visualize the structures in the field of structural biology. 

It is possible to use PyMOL via the Python interface to create visualizations directly through Python. These visualizations can then be viewed within the PyMOL editor. 
The purpose of this notebook is to provide help on how to use PyMOL via the Python interface. 

A video introduction on how to use the PyMOL editor is available in the Moodle course. Additional resources to assist you with PyMOL can be found at https://sites.pitt.edu/~epolinko/IntroPyMOL.pdf and https://pymolwiki.org/index.php/Main_Page.

```python
# Import the pymol module
from pymol.cgo import cmd

import pandas as pd
```

### Example 1:
1. Reinitialize so that you get an empty session 
3. Setting the background color
2. Fetching a protein from pdb
3. Show the protein as cartoon
4. Change to color of the protein
5. Select the water molecules
6. Remove the water molecules sourounding the protein (makes the visualization look prettier)
5. Save the pymol session

The exported file can now be opend via pymol.

```python
protein_name = "1a0j"
output_file = "example1.pse"

cmd.reinitialize()                          # Reinitialize the pymol session
cmd.bg_color("black")                       # Set the background color to black (black is the default, just for demonstration purposes)   
cmd.fetch(protein_name)                     # Fetch the protein from the RCSB PDB
cmd.show("cartoon", protein_name)           # Show the protein as cartoon
cmd.color("wheat", protein_name)            # Color the protein in wheat  
cmd.select("waters", "resn HOH")            # Select all water molecules (first parameter is the name of the selection, second parameter is the selection criteria)
cmd.remove("waters")                        # Remove all water molecules
cmd.delete("waters")                        # Delete the selection object
cmd.save(output_file)                       # Save the visualizationas pymol script (fileextension .pse)
```

### Example 2:
This example uses the files from the challange. It should give you a feeling how the bindingsites/pockets you have to predict in the challenge look.

1. Reinitialize so that you get an empty session 
2. Load protein from disk
3. Show the protein as surface
4. Change to color of the protein
5. Select certain residues from the protein
6. Create a new object from the selection
7. Change the color of the selection object
8. Save the pymol session

```python
protein_id = 1821
protein_path = f"../data/train/{protein_id}_protein.pdb"
residue_info = "../data/train/train.csv"
output_file = "example2.pse"

# Load the train.csv from the challenge
df_residue_info = pd.read_csv(residue_info)
residues = df_residue_info.loc[lambda x: x["id"] == protein_id, "resid"].item().split(" ")
```

```python
cmd.reinitialize()                          # Reinitialize the pymol session    
cmd.load(protein_path, protein_id)          # Load the protein from the file
cmd.show("surface", protein_id)             # Show the protein as surface
cmd.color("wheat", protein_id)              # Color the protein in wheat


selection = ""                              # Create a selection string from the residues, e.g. "chain A and resi 1 or chain A and resi 2 or chain A and resi 3"
for i, resid in enumerate(residues):
    selection += f"chain {resid.split('_')[0]} and resi {resid.split('_')[1]}"
    if i < len(residues) - 1:
        selection += " or "                 # Add "or" if it is not the last residue
if selection != "":
    cmd.select("s", selection)              # Select the residues created by selection
    cmd.show("surface", "s")                # Show the selection as surface
    cmd.create("pocket", "s")               # Create a new object from the selection
    cmd.color("green", "pocket")            # Color the new object in green
    cmd.delete("s")                         # Delete the selection (not the object)
                             
cmd.save(output_file)                       # Save the visualizationas pymol script (fileextension .pse)
```
