# Exercise 1: Simulation des zentralen Grenzwertsatzes

## 1. Übersetzung der Aufgabenstellung

Der zentrale Grenzwertsatz (Central Limit Theorem, CLT) besagt:

Sind $z_1,\dots,z_l$ unabhängig und identisch verteilt mit Erwartungswert $\mu$ und endlicher Varianz $\sigma^2$, dann gilt für den Stichprobenmittelwert

$$
\hat{\mu}=\frac{1}{l}\sum_{i=1}^{l}z_i:
$$

$$
\sqrt{l}(\hat{\mu}-\mu)
\xrightarrow{d}
\mathcal{N}(0,\sigma^2)
\qquad \text{für } l\to\infty.
$$

Das gilt unabhängig von der ursprünglichen Verteilung, sofern die genannten Voraussetzungen erfüllt sind.

Äquivalent dazu ist der standardisierte Stichprobenmittelwert, der sogenannte z-Score,

$$
Z_l=\frac{\hat{\mu}-\mu}{\sigma/\sqrt{l}},
$$

für große Stichproben näherungsweise standardnormalverteilt:

$$
Z_l\approx\mathcal{N}(0,1).
$$

Deshalb werden Schätzer wie der Stichprobenmittelwert in der Praxis häufig als näherungsweise normalverteilt behandelt. Dafür muss die Stichprobe aber ausreichend groß sein.

Ziel dieser Aufgabe ist es, zu untersuchen, wie gut diese Näherung bereits bei endlichen Stichprobengrößen funktioniert.

**Exponentialverteilung:**

Für die Rate $\lambda=1$ lautet die Dichte

$$
p(z)=e^{-z},\qquad z\geq 0.
$$

Erwartungswert und Varianz sind

$$
\mathbb{E}[z]=1,\qquad \operatorname{Var}(z)=1.
$$

**Pareto-Verteilung:**

Für den Skalenparameter $x_m=1$ und $\alpha=2.1$ lautet die Dichte

$$
p(z)=\alpha z^{-\alpha-1},\qquad z\geq 1.
$$

Erwartungswert und Varianz sind

$$
\mathbb{E}[z]=\frac{\alpha}{\alpha-1},
$$

$$
\operatorname{Var}(z)
=\frac{\alpha}{(\alpha-1)^2(\alpha-2)}.
$$

Weil $\alpha>2$ ist, ist die Varianz endlich und der zentrale Grenzwertsatz anwendbar.

Stichproben lassen sich mit `rng.pareto(alpha, n) + 1` erzeugen. Das `+ 1` ist notwendig, weil NumPy die um $-1$ verschobene Pareto-Verteilung, die Lomax-Verteilung, erzeugt.

Der zentrale Grenzwertsatz sagt: Je größer die Stichproben werden, desto ähnlicher wird die Verteilung der standardisierten Durchschnittswerte einer Normalverteilung.

Die Standardisierung erfolgt mit

$$
Z_l=\frac{\hat{\mu}-\mu}{\sigma/\sqrt{l}}.
$$

Dabei bedeutet:

- $\hat{\mu}$: der Durchschnitt einer gezogenen Stichprobe,
- $\mu$: der tatsächliche Erwartungswert der ursprünglichen Verteilung,
- $\sigma$: deren tatsächliche Standardabweichung,
- $\sigma/\sqrt{l}$: die Standardabweichung des Stichprobenmittelwerts.

Wir ziehen also den tatsächlichen Erwartungswert ab und teilen durch die typische Schwankungsbreite des Durchschnitts.

**So liest du die Ergebnisse:**

- **Dichteplot:** Das Histogramm sollte ungefähr der roten Glockenkurve folgen.
- **Kennzahlen:** Für eine Standardnormalverteilung sind Mittelwert $0$, Varianz $1$, Schiefe $0$ und Exzess-Kurtosis $0$.

Die Funktion `stats.kurtosis` verwendet standardmäßig die Exzess-Kurtosis. Daher ist hier der Vergleichswert $0$ und nicht $3$.

Die vorhandenen Hilfsfunktionen und `sample_sizes` müssen zuvor ausgeführt worden sein.

```python
# Exponential distribution with rate lambda = 1
# NumPy uses scale = 1 / lambda.

exp_mean = 1.0
exp_var = 1.0

exp_sample_means = simulate_sample_means(
    sampler=lambda n: rng.exponential(scale=1.0, size=n),
    sample_sizes=sample_sizes,
    monte_carlo_sims=2000,
)

exp_z = z_scores(
    exp_sample_means,
    mean=exp_mean,
    var=exp_var,
)

plot_results(exp_z, "Exponential distribution (lambda=1)")
print_moments(exp_z)
```

### Pareto-Verteilung

Für $\alpha=2.1$ erhalten wir

$$
\mu=\frac{2.1}{1.1}\approx 1.9091
$$

und

$$
\sigma^2
=\frac{2.1}{1.1^2\cdot 0.1}
\approx 17.3554.
$$

```python
# Classical Pareto distribution with x_m = 1 and alpha = 2.1

alpha = 2.1

pareto_mean = alpha / (alpha - 1)
pareto_var = alpha / ((alpha - 1)**2 * (alpha - 2))

pareto_sample_means = simulate_sample_means(
    sampler=lambda n: rng.pareto(alpha, size=n) + 1,
    sample_sizes=sample_sizes,
    monte_carlo_sims=2000,
)

pareto_z = z_scores(
    pareto_sample_means,
    mean=pareto_mean,
    var=pareto_var,
)

plot_results(pareto_z, "Pareto distribution (alpha=2.1)")
print_moments(pareto_z)
```

## 4. Lösung zu 1b: Interpretation

### Exponentialverteilung

Bei kleinen Stichproben, insbesondere $l=2$ und $l=5$, sind die standardisierten Mittelwerte noch deutlich rechtsschief. Im QQ-Plot zeigen sich Abweichungen von der Geraden, und das Histogramm ist noch nicht symmetrisch.

Bei $l=20$ ist die Normalverteilungsnäherung für den zentralen Bereich bereits brauchbar. In den Randbereichen können weiterhin erkennbare Abweichungen auftreten.

Bei $l=100$ ist die Näherung normalerweise deutlich besser und bei $l=1000$ meist sehr gut. Kleine Unregelmäßigkeiten bleiben wegen der endlichen Anzahl an Simulationswiederholungen möglich.

Das lässt sich auch theoretisch nachvollziehen: Für die Exponentialverteilung mit Rate $1$ haben die standardisierten Mittelwerte die Schiefe

$$
\operatorname{Skew}(Z_l)=\frac{2}{\sqrt{l}}
$$

und die Exzess-Kurtosis

$$
\operatorname{Kurt}_{\mathrm{excess}}(Z_l)=\frac{6}{l}.
$$

Beide Größen gehen mit wachsender Stichprobengröße gegen $0$, wie bei einer Normalverteilung.

### Pareto-Verteilung

Die Pareto-Verteilung mit $\alpha=2.1$ besitzt einen schweren rechten Verteilungsschwanz: Sehr große Werte sind selten, aber wesentlich wahrscheinlicher als bei der Exponentialverteilung.

Ein einzelner extremer Wert kann dadurch den Durchschnitt einer ganzen Stichprobe stark nach oben ziehen.

Für $l=2$ und $l=5$ ist die Normalverteilungsnäherung schlecht. Auch bei $l=20$ und $l=100$ sind deutliche Abweichungen zu erwarten. Selbst bei $l=1000$ ist eine gute Näherung, insbesondere in den Randbereichen, noch nicht zuverlässig gewährleistet.

Der zentrale Grenzwertsatz gilt trotzdem, weil die Varianz endlich ist. Er garantiert jedoch keine gute Näherung bei einer bestimmten endlichen Stichprobengröße.

Die Konvergenz ist hier besonders langsam, weil $\alpha=2.1$ nur knapp über der Grenze $\alpha=2$ für eine endliche Varianz liegt. Außerdem sind das dritte und vierte Moment nicht endlich. Daher besitzen auch die standardisierten Stichprobenmittelwerte bei endlichem $l$ keine endliche theoretische Schiefe oder Kurtosis.

Die im Programm berechneten Werte für Schiefe und Kurtosis sind empirische Kennzahlen der Simulation. Sie können bei der Pareto-Verteilung stark schwanken und sollten nicht als stabile Schätzungen endlicher theoretischer Größen interpretiert werden.

### Warum die empirische Varianz von 1 abweichen kann

Für beide Verteilungen gilt bei jeder Stichprobengröße exakt:

$$
\mathbb{E}[Z_l]=0,
\qquad
\operatorname{Var}(Z_l)=1.
$$

Das allein bedeutet noch nicht, dass $Z_l$ normalverteilt ist.

Die ausgegebenen Kennzahlen werden nur aus 2000 simulierten Mittelwerten berechnet. Besonders bei der Pareto-Verteilung kann die empirische Varianz deutlich von $1$ abweichen: Seltene Extremwerte tragen stark zur theoretischen Varianz bei, werden in einer endlichen Simulation aber möglicherweise gar nicht oder nur vereinzelt beobachtet.

### Konsequenzen für die Praxis

Die Exponentialverteilung nähert sich wesentlich schneller einer Normalverteilung an als die betrachtete Pareto-Verteilung.

Es gibt daher keine allgemeingültige Regel, dass beispielsweise 30 Beobachtungen immer für den zentralen Grenzwertsatz ausreichen. Wie viele Beobachtungen nötig sind, hängt von der ursprünglichen Verteilung und der gewünschten Genauigkeit ab.

Ein normalverteilungsbasiertes Konfidenzintervall der Form

$$
\hat{\mu}\pm 1.96\frac{\sigma}{\sqrt{l}}
$$

kann bei ausreichend großen Stichproben ungefähr 95 % Überdeckungswahrscheinlichkeit erreichen. Bei schweren Verteilungsschwänzen kann die tatsächliche Überdeckungswahrscheinlichkeit bei endlichem $l$ jedoch erheblich davon abweichen.

Wird die unbekannte Standardabweichung zusätzlich aus den Daten geschätzt, können seltene Extremwerte auch diese Schätzung stark beeinflussen.

Für praktische Anwendungen sollte deshalb geprüft werden, ob starke Schiefe oder schwere Verteilungsschwänze vorliegen. Eine endliche Varianz sichert die asymptotische Gültigkeit des zentralen Grenzwertsatzes, aber keine schnelle oder bei einer konkreten Stichprobengröße bereits hinreichend genaue Normalverteilungsnäherung.

## 1. Übersetzung der Aufgabenstellung

Für einen Schätzer gilt die Bias-Varianz-Zerlegung:

$$
\operatorname{MSE}(\hat w)
=
\operatorname{Var}(\hat w)+b(\hat w)^2,
$$

wobei

$$
b(\hat w)=\mathbb E[\hat w]-w
$$

der Bias ist.

In dieser Aufgabe sind die Beobachtungen unabhängig und identisch normalverteilt:

$$
z_1,\dots,z_l\overset{\mathrm{i.i.d.}}{\sim}
\mathcal N(\mu,\sigma^2).
$$

### 2a) Optimale Skalierung des Stichprobenmittelwerts

$$
\hat\mu_\gamma
=
\frac{\gamma}{l}\sum_{i=1}^{l}z_i.
$$

Aus der Vorlesung ist bekannt:

$$
\operatorname{MSE}(\hat\mu_\gamma)
=
\frac{\gamma^2\sigma^2}{l}
+
\mu^2(\gamma-1)^2.
$$

1. Bestimme den Faktor $\gamma^\ast$, der den MSE minimiert. Zeige, dass dieser Schätzer einen kleineren MSE hat als der unverzerrte und effiziente Stichprobenmittelwert $\hat\mu_1$.
2. Warum ist dieser optimale Schätzer trotzdem praktisch nicht verwendbar? Wie verhält sich $\gamma^\ast$ für $l\to\infty$?

### 2b) Verkleinerung des Varianzschätzers

Definiere

$$
\hat\mu=\frac1l\sum_{i=1}^{l}z_i,
\qquad
S=\sum_{i=1}^{l}(z_i-\hat\mu)^2.
$$

Es gilt:

$$
\mathbb E[S]=(l-1)\sigma^2,
$$

$$
\operatorname{Var}(S)=2(l-1)\sigma^4.
$$

Untersuche die Familie von Varianzschätzern

$$
\hat\sigma_c^2=cS,\qquad c>0.

$$
\frac{2\sigma^4}{l}
$$

für unverzerrte Schätzer von $\sigma^2$.

Führe den Vergleich allgemein und für $l=10$, $\sigma^2=1$ durch. Welche Schätzer haben eine Varianz unterhalb der Schranke, und warum ist das kein Widerspruch?

## 2. Einfache Erklärung

Ein Schätzer versucht, aus einer begrenzten Stichprobe einen unbekannten Wert zu bestimmen.

Dabei unterscheiden wir:

- **Bias:** Liegt der Schätzer im Durchschnitt über oder unter dem richtigen Wert?
- **Varianz:** Wie stark schwankt der Schätzer zwischen verschiedenen Stichproben?
- **MSE:** Wie groß ist sein durchschnittlicher quadratischer Abstand zum richtigen Wert?

Es gilt:

$$
\text{MSE}=\text{Varianz}+\text{Bias}^2.
$$

Ein unverzerrter Schätzer trifft den richtigen Wert im Durchschnitt. Trotzdem kann er stark schwanken.

Manchmal lohnt es sich deshalb, eine kleine systematische Abweichung zuzulassen, wenn dadurch die Schwankungen ausreichend kleiner werden. Dann sinkt der gesamte MSE.

Genau das untersuchen wir hier:

- In **2a** multiplizieren wir den Stichprobenmittelwert mit einem Faktor und ziehen ihn dadurch Richtung null.
- In **2b** multiplizieren wir die Summe der quadrierten Abweichungen mit verschiedenen Faktoren.
- In **2c** überprüfen wir die theoretischen Ergebnisse durch Simulation.

## 3. Lösung zu 2a

### Schritt 1: Den MSE minimieren

Gegeben ist

$$
\operatorname{MSE}(\hat\mu_\gamma)
=
\frac{\gamma^2\sigma^2}{l}
+
\mu^2(\gamma-1)^2.
$$

Wir leiten nach $\gamma$ ab:

$$
\frac{d}{d\gamma}\operatorname{MSE}
=
2\gamma\frac{\sigma^2}{l}
+
2\mu^2(\gamma-1).
$$

Für ein Minimum setzen wir die Ableitung gleich null:

$$
\gamma\frac{\sigma^2}{l}
+
\gamma\mu^2-\mu^2=0.
$$

Daraus folgt

$$
\gamma\left(\mu^2+\frac{\sigma^2}{l}\right)=\mu^2
$$

und damit

$$
\boxed{
\gamma^\ast
=
\frac{\mu^2}{\mu^2+\sigma^2/l}
=
\frac{l\mu^2}{l\mu^2+\sigma^2}
}.
$$

Für $\sigma^2>0$ ist die zweite Ableitung positiv:

$$
\frac{d^2}{d\gamma^2}\operatorname{MSE}
=
2\frac{\sigma^2}{l}+2\mu^2>0.
$$

Es handelt sich daher um das eindeutige Minimum.

### Schritt 2: Den minimalen MSE bestimmen

Zur Vereinfachung setzen wir

$$
A=\frac{\sigma^2}{l},
\qquad
B=\mu^2.
$$

Dann gilt

$$
\gamma^\ast=\frac{B}{A+B},
\qquad
\gamma^\ast-1=-\frac{A}{A+B}.
$$

Einsetzen ergibt

$$
\operatorname{MSE}(\hat\mu_{\gamma^\ast})
=
A\frac{B^2}{(A+B)^2}
+
B\frac{A^2}{(A+B)^2}
=
\frac{AB}{A+B}.
$$

Somit ist

$$
\boxed{
\operatorname{MSE}(\hat\mu_{\gamma^\ast})
=
\frac{\mu^2\sigma^2}{l\mu^2+\sigma^2}
}.
$$

Der gewöhnliche Stichprobenmittelwert hat den MSE

$$
\operatorname{MSE}(\hat\mu_1)=\frac{\sigma^2}{l}=A.
$$

Die Differenz beträgt

$$
A-\frac{AB}{A+B}
=
\frac{A^2}{A+B}>0.
$$

Daher gilt für $\sigma^2>0$:

$$
\boxed{
\operatorname{MSE}(\hat\mu_{\gamma^\ast})
<
\operatorname{MSE}(\hat\mu_1)
}.
$$

Für $\mu\neq0$ ist der optimale Schätzer verzerrt, denn

$$
b(\hat\mu_{\gamma^\ast})
=
(\gamma^\ast-1)\mu\neq0.
$$

Seine kleinere Varianz gleicht den zusätzlichen quadrierten Bias mehr als aus.

### Schritt 3: Warum ist der optimale Schätzer nicht direkt verwendbar?

Der optimale Faktor

$$
\gamma^\ast=\frac{l\mu^2}{l\mu^2+\sigma^2}
$$

hängt vom unbekannten Parameter $\mu$ ab, den wir eigentlich erst schätzen wollen. Falls auch $\sigma^2$ unbekannt ist, besteht zusätzlich eine Abhängigkeit von diesem unbekannten Parameter.

Der Faktor lässt sich deshalb nicht direkt aus bekannten Größen berechnen.

Man könnte unbekannte Größen durch Schätzungen ersetzen. Das ergibt jedoch einen anderen Schätzer, für den die gerade hergeleitete MSE-Garantie nicht automatisch gilt.

### Schritt 4: Verhalten für große Stichproben

Für einen festen Wert $\mu\neq0$ gilt:

$$
\boxed{
\gamma^\ast\longrightarrow1
\qquad\text{für }l\to\infty
}.
$$

Bei großen Stichproben wird daher immer weniger geschrumpft.

Der Grund: Die Varianz des gewöhnlichen Stichprobenmittelwerts sinkt bereits wie $\sigma^2/l$. Eine zusätzliche Verkleinerung ist immer weniger vorteilhaft.

**Sonderfall $\mu=0$:**

Dann gilt $\gamma^\ast=0$ für alle $l$. Der entsprechende theoretische Schätzer ist konstant null und hat MSE null. Diese Wahl setzt allerdings voraus, dass wir den wahren Mittelwert bereits kennen.

## 4. Lösung zu 2b

Wir nehmen $l>1$ an und betrachten

$$
\hat\sigma_c^2=cS.
$$

### Schritt 1: Bias

Aus

$$
\mathbb E[S]=(l-1)\sigma^2
$$

folgt

$$
\mathbb E[\hat\sigma_c^2]
=
c(l-1)\sigma^2.
$$

Somit ist der Bias

$$
\boxed{
b(\hat\sigma_c^2)
=
\big(c(l-1)-1\big)\sigma^2
}.
$$

### Schritt 2: Varianz

Für eine konstante Skalierung gilt

$$
\operatorname{Var}(cS)=c^2\operatorname{Var}(S).
$$

Damit erhalten wir

$$
\boxed{
\operatorname{Var}(\hat\sigma_c^2)
=
2c^2(l-1)\sigma^4
}.
$$

Hier bezeichnet $\sigma^4=(\sigma^2)^2$ das Quadrat der wahren Varianz.

### Schritt 3: MSE

Mit der Bias-Varianz-Zerlegung folgt

$$
\boxed{
\operatorname{MSE}(\hat\sigma_c^2)
=
\sigma^4
\left[
2c^2(l-1)
+
\big(c(l-1)-1\big)^2
\right]
}.
$$

Ausmultiplizieren ergibt

$$
\operatorname{MSE}(\hat\sigma_c^2)
=
\sigma^4
\left[
(l-1)(l+1)c^2-2(l-1)c+1
\right].
$$

### Schritt 4: Optimalen Faktor bestimmen

Wir leiten nach $c$ ab:

$$
\frac{d}{dc}\operatorname{MSE}
=
2\sigma^4(l-1)\big((l+1)c-1\big).
$$

Die Ableitung ist null für

$$
\boxed{
c^\ast=\frac1{l+1}
}.
$$

Die zweite Ableitung ist positiv:

$$
\frac{d^2}{dc^2}\operatorname{MSE}
=
2\sigma^4(l-1)(l+1)>0.
$$

Der MSE-optimale Schätzer innerhalb der betrachteten Familie lautet also

$$
\boxed{
\hat\sigma_{\mathrm{opt}}^2
=
\frac{S}{l+1}
}.
$$

Anders als der optimale Faktor aus 2a hängt $c^\ast$ nur von der bekannten Stichprobengröße ab. Dieser Schätzer ist daher direkt verwendbar.

### Schritt 5: Allgemeiner Vergleich

| Schätzer | Faktor $c$ | Bias | Varianz | MSE |
|---|---|---|---|---|
| MLE | $\frac1l$ | $-\frac{\sigma^2}{l}$ | $\frac{2(l-1)}{l^2}\sigma^4$ | $\frac{2l-1}{l^2}\sigma^4$ |
| Unverzerrt | $\frac1{l-1}$ | $0$ | $\frac{2}{l-1}\sigma^4$ | $\frac{2}{l-1}\sigma^4$ |
| MSE-optimal | $\frac1{l+1}$ | $-\frac{2\sigma^2}{l+1}$ | $\frac{2(l-1)}{(l+1)^2}\sigma^4$ | $\frac{2}{l+1}\sigma^4$ |

Die angegebene Cramér–Rao-Schranke für unverzerrte Schätzer beträgt

$$
\operatorname{CRLB}=\frac{2\sigma^4}{l}.
$$

Für die Varianzen gilt:

$$
\boxed{
\operatorname{Var}(\hat\sigma_{\mathrm{opt}}^2)
<
\operatorname{Var}(\hat\sigma_{\mathrm{MLE}}^2)
<
\operatorname{CRLB}
<
\operatorname{Var}(\bar\sigma^2)
}.
$$

Auch die MSEs erfüllen:

$$
\boxed{
\operatorname{MSE}(\hat\sigma_{\mathrm{opt}}^2)
<
\operatorname{MSE}(\hat\sigma_{\mathrm{MLE}}^2)
<
\operatorname{CRLB}
<
\operatorname{MSE}(\bar\sigma^2)
}.
$$

Der unverzerrte Schätzer erreicht die angegebene Schranke bei endlichem $l$ also nicht.

### Schritt 6: Numerischer Vergleich für $l=10$ und $\sigma^2=1$

Hier ist $\sigma^4=1$ und

$$
\operatorname{CRLB}=\frac2{10}=0.2.
$$

| Schätzer | Bias | Varianz | MSE |
|---|---:|---:|---:|
| MLE: $S/10$ | −0.100000 | 0.180000 | 0.190000 |
| Unverzerrt: $S/9$ | 0.000000 | 0.222222 | 0.222222 |
| MSE-optimal: $S/11$ | −0.181818 | 0.148760 | 0.181818 |

Der MSE-optimale Schätzer unterschätzt die wahre Varianz im Durchschnitt stärker als der MLE. Dafür schwankt er weniger. Insgesamt besitzt er den kleinsten MSE.

### Warum ist das kein Widerspruch zur Cramér–Rao-Schranke?

Die hier verwendete Schranke

$$
\operatorname{Var}(\hat\sigma^2)\geq\frac{2\sigma^4}{l}
$$

gilt für **unverzerrte** Schätzer.

MLE und MSE-optimaler Schätzer sind beide verzerrt. Ihre Varianzen dürfen deshalb unter dieser Schranke liegen.

Auch ihre MSEs dürfen kleiner als dieser Vergleichswert sein: Die angegebene Schranke ist keine allgemeine MSE-Untergrenze für verzerrte Schätzer.

Der folgende Code erzeugt 10 000 unabhängige Stichproben. Jede Zeile enthält eine Stichprobe aus 10 standardnormalverteilten Beobachtungen.

Für jede Stichprobe werden die drei Varianzschätzer berechnet. Anschließend vergleichen wir empirischen Bias, empirische Varianz und empirischen MSE mit den theoretischen Werten.

```python
import numpy as np

rng_ex2 = np.random.default_rng(0)

n = 10
trials = 10_000
true_mean = 0.0
true_var = 1.0

# Each row is one independent sample of size n.
z = rng_ex2.normal(
    loc=true_mean,
    scale=np.sqrt(true_var),
    size=(trials, n),
)

sample_means = np.mean(z, axis=1, keepdims=True)
S = np.sum((z - sample_means)**2, axis=1)

factors = {
    "MLE": 1 / n,
    "Unbiased": 1 / (n - 1),
    "MSE-optimal": 1 / (n + 1),
}

crlb = 2 * true_var**2 / n

print(f"CRLB for unbiased estimators: {crlb:.6f}\n")

print(
    f"{'Estimator':<14}"
    f"{'Bias emp.':>12}"
    f"{'Bias theory':>14}"
    f"{'Var emp.':>12}"
    f"{'Var theory':>14}"
    f"{'MSE emp.':>12}"
    f"{'MSE theory':>14}"
)

for name, c in factors.items():
    estimates = c * S

    # Empirical quantities across the 10,000 trials.
    empirical_bias = np.mean(estimates) - true_var
    empirical_var = np.var(estimates, ddof=0)
    empirical_mse = np.mean((estimates - true_var)**2)

    # Theoretical quantities from Exercise 2b.
    theoretical_bias = (c * (n - 1) - 1) * true_var
    theoretical_var = 2 * c**2 * (n - 1) * true_var**2
    theoretical_mse = theoretical_var + theoretical_bias**2

    print(
        f"{name:<14}"
        f"{empirical_bias:>12.6f}"
        f"{theoretical_bias:>14.6f}"
        f"{empirical_var:>12.6f}"
        f"{theoretical_var:>14.6f}"
        f"{empirical_mse:>12.6f}"
        f"{theoretical_mse:>14.6f}"
    )
```

### Interpretation der Simulation

Die empirischen Ergebnisse sollten ungefähr mit den theoretischen Werten übereinstimmen. Kleine Abweichungen sind aufgrund der endlichen Anzahl von 10 000 Wiederholungen zu erwarten.

- Der unverzerrte Schätzer hat einen Bias nahe null, aber die größte Varianz und den größten MSE der drei Schätzer.
- Der MLE unterschätzt die wahre Varianz im Durchschnitt etwas. Seine geringere Varianz führt dennoch zu einem kleineren MSE als beim unverzerrten Schätzer.
- Der MSE-optimale Schätzer unterschätzt die wahre Varianz stärker, hat dafür aber die kleinste Varianz und den kleinsten theoretischen MSE der drei Schätzer.
- MLE und MSE-optimaler Schätzer können unter der angegebenen Cramér–Rao-Schranke liegen, weil sie verzerrt sind.

Die Aufgabe zeigt: Ein kleiner Bias kann vorteilhaft sein, wenn er die Varianz ausreichend reduziert. Ein unverzerrter Schätzer ist deshalb nicht automatisch der beste Schätzer bezüglich des MSE.

### 2b) Shrinking the variance estimator

For $\hat\sigma_c^2=cS$, using $\mathbb E[S]=(l-1)\sigma^2$ and $\operatorname{Var}(S)=2(l-1)\sigma^4$, we obtain

$$
b(\hat\sigma_c^2)=[c(l-1)-1]\sigma^2,
$$

$$
\operatorname{Var}(\hat\sigma_c^2)=2c^2(l-1)\sigma^4,
$$

$$
\operatorname{MSE}(\hat\sigma_c^2)
=\sigma^4\left[2c^2(l-1)+(c(l-1)-1)^2\right].
$$

Differentiating with respect to $c$ gives

$$
\frac{d}{dc}\operatorname{MSE}
=2\sigma^4(l-1)[(l+1)c-1]=0.
$$

Thus, for $l>1$, the unique minimum is attained at

$$
\boxed{c^\ast=\frac1{l+1}}.
$$

| Estimator | Bias | Variance | MSE |
|---|---|---|---|
| MLE: $S/l$ | $-\sigma^2/l$ | $2(l-1)\sigma^4/l^2$ | $(2l-1)\sigma^4/l^2$ |
| Unbiased: $S/(l-1)$ | $0$ | $2\sigma^4/(l-1)$ | $2\sigma^4/(l-1)$ |
| MSE-optimal: $S/(l+1)$ | $-2\sigma^2/(l+1)$ | $2(l-1)\sigma^4/(l+1)^2$ | $2\sigma^4/(l+1)$ |

For $l=10$ and $\sigma^2=1$, the CRLB is $2\sigma^4/l=0.2$:

| Estimator | Bias | Variance | MSE |
|---|---:|---:|---:|
| MLE | −0.1000 | 0.1800 | 0.1900 |
| Unbiased | 0.0000 | 0.2222 | 0.2222 |
| MSE-optimal | −0.1818 | 0.1488 | 0.1818 |

For all $l>1$, both variance and MSE follow the ordering

$$
\text{MSE-optimal}<\text{MLE}<\text{CRLB}<\text{unbiased}.
$$

This is no contradiction: the stated CRLB applies only to unbiased estimators. The two biased estimators reduce variance enough to achieve a smaller MSE.

## Aufgabe 3: Fisher-Information und Cramér–Rao-Schranke

### Grundidee

Die **Fisher-Information** beschreibt, wie viel eine Beobachtung über einen unbekannten Parameter verrät.

Die **Cramér–Rao-Schranke (CRLB)** gibt eine Untergrenze für die Varianz unverzerrter Schätzer an:

$$
\operatorname{Var}(\hat w)\geq\frac{1}{I_F^{(l)}(w)}.
$$

Ein Schätzer heißt **unverzerrt**, wenn er im Durchschnitt den richtigen Parameterwert liefert. Er heißt **effizient**, wenn er zusätzlich die CRLB erreicht.

### 3a) Information einer gesamten Stichprobe

**Übersetzung der Aufgabe:**

Zeige, dass die Fisher-Information von $l$ unabhängigen, identisch verteilten Beobachtungen gleich dem $l$-Fachen der Information einer einzelnen Beobachtung ist. Vergleiche dies mit den Formeln aus der Vorlesung für normalverteilte Daten.

**Einfach erklärt:**

Wenn eine Beobachtung eine bestimmte Menge an Information liefert, liefern zehn unabhängige Beobachtungen zehnmal so viel Information.

**Lösung:**

Wegen der Unabhängigkeit ist die Likelihood ein Produkt:

$$
\mathcal L(\mathbf Z;w)=\prod_{i=1}^{l}p(z_i;w).
$$

Der Logarithmus macht daraus eine Summe:

$$
\log\mathcal L(\mathbf Z;w)
=
\sum_{i=1}^{l}\log p(z_i;w).
$$

Damit gilt:

$$
I_F^{(l)}(w)
=
-\mathbb E\left[
\sum_{i=1}^{l}
\frac{\partial^2}{\partial w^2}\log p(z_i;w)
\right]
=
lI_F(w).
$$

Also:

$$
\boxed{I_F^{(l)}(w)=lI_F(w)}.
$$

Für normalverteilte Daten folgt:

$$
I_F^{(l)}(\mu)=\frac{l}{\sigma^2},
\qquad
I_F^{(l)}(\sigma^2)=\frac{l}{2\sigma^4}.
$$

Die entsprechenden Formeln der Vorlesung beziehen sich somit auf die gesamte Stichprobe.

**Bedeutung:** Mehr unabhängige Daten erhöhen die Information und senken die CRLB.

### 3b) Fisher-Information der Exponentialrate

**Übersetzung der Aufgabe:**

Berechne die Fisher-Information für die Rate $\lambda$ einer Exponentialverteilung. Verwende zuerst die zweite Ableitung der Log-Dichte und überprüfe das Ergebnis anschließend mit dem quadrierten Score. Bestimme die CRLB für $l$ Beobachtungen.

**Einfach erklärt:**

Bei der Exponentialverteilung beschreibt $\lambda$, wie schnell Ereignisse eintreten. Der mittlere Abstand zwischen Ereignissen ist $1/\lambda$.

Wir untersuchen, wie genau sich $\lambda$ aus beobachteten Wartezeiten schätzen lässt.

**Lösung:**

Die Dichte lautet:

$$
p(z;\lambda)=\lambda e^{-\lambda z}.
$$

Ihre Log-Dichte ist:

$$
\log p(z;\lambda)=\log\lambda-\lambda z.
$$

Die erste Ableitung ist der Score:

$$
s(z;\lambda)=\frac1\lambda-z.
$$

Die zweite Ableitung lautet:

$$
\frac{\partial^2}{\partial\lambda^2}\log p(z;\lambda)
=
-\frac1{\lambda^2}.
$$

Daher:

$$
\boxed{I_F(\lambda)=\frac1{\lambda^2}}.
$$

Die Überprüfung mit dem Score ergibt dasselbe:

$$
\mathbb E[s(z;\lambda)^2]
=
\mathbb E\left[\left(z-\frac1\lambda\right)^2\right]
=
\operatorname{Var}(z)
=
\frac1{\lambda^2}.
$$

Für $l$ Beobachtungen gilt:

$$
I_F^{(l)}(\lambda)=\frac l{\lambda^2}.
$$

Damit lautet die CRLB:

$$
\boxed{
\operatorname{Var}(\hat\lambda)\geq\frac{\lambda^2}{l}
}
$$

für unverzerrte Schätzer.

**Bedeutung:** Bei mehr Beobachtungen wird eine genauere Schätzung der Rate möglich.

### 3c) Schätzung des Mittelwerts statt der Rate

**Übersetzung der Aufgabe:**

Schreibe die Exponentialverteilung mit ihrem Mittelwert $\theta=1/\lambda$ als Parameter.

1. Berechne die Fisher-Information für $\theta$.
2. Zeige, wie sich die Fisher-Information bei einer Parametertransformation verändert.
3. Bestimme die CRLB und zeige, dass der Stichprobenmittelwert effizient ist. Ist sein Kehrwert ebenfalls ein unverzerrter und effizienter Schätzer der Rate?

**Einfach erklärt:**

Wir beschreiben dieselbe Verteilung auf zwei Arten:

- $\lambda$: Rate der Ereignisse,
- $\theta=1/\lambda$: durchschnittliche Wartezeit.

Der Durchschnitt der beobachteten Wartezeiten ist ein sehr guter Schätzer für $\theta$. Sein Kehrwert ist jedoch nicht automatisch genauso gut für $\lambda$.

**Lösung:**

Mit $\theta=1/\lambda$ lautet die Log-Dichte:

$$
\log p(z;\theta)=-\log\theta-\frac z\theta.
$$

Die zweite Ableitung ist:

$$
\frac{\partial^2}{\partial\theta^2}\log p(z;\theta)
=
\frac1{\theta^2}-\frac{2z}{\theta^3}.
$$

Mit $\mathbb E[z]=\theta$ folgt:

$$
I_F(\theta)
=
-\left(
\frac1{\theta^2}-\frac{2\theta}{\theta^3}
\right)
=
\boxed{\frac1{\theta^2}}.
$$

**Allgemeine Transformationsregel:**

Für $w=g(\theta)$ ergibt die Kettenregel:

$$
s_\theta=s_w\,g'(\theta).
$$

Quadrieren und Erwartungswert bilden liefert:

$$
\boxed{
I_{F,\theta}(\theta)
=
I_{F,w}(g(\theta))[g'(\theta)]^2
}.
$$

Für $\lambda=1/\theta$ gilt:

$$
I_F(\theta)
=
\frac1{\lambda^2}
\left(-\frac1{\theta^2}\right)^2
=
\frac1{\theta^2}.
$$

Die CRLB für $l$ Beobachtungen ist:

$$
\boxed{
\operatorname{Var}(\hat\theta)\geq\frac{\theta^2}{l}
}.
$$

Der Stichprobenmittelwert erfüllt:

$$
\mathbb E[\hat\mu]=\theta,
\qquad
\operatorname{Var}(\hat\mu)=\frac{\theta^2}{l}.
$$

Er ist somit **unverzerrt und effizient**.

Sein Kehrwert $1/\hat\mu$ ist dagegen nach oben verzerrt: Kleine beobachtete Mittelwerte erzeugen besonders große Kehrwerte. Diese gleichen sich nicht symmetrisch mit den Kehrwerten großer Mittelwerte aus.

**Zusammenfassung von Aufgabe 3:**

Mehr unabhängige Beobachtungen liefern mehr Information. Der Stichprobenmittelwert schätzt die mittlere Wartezeit effizient. Durch Kehrwertbildung geht die Unverzerrtheit jedoch verloren.

## Aufgabe 4: Maximum-Likelihood-Schätzung

### Grundidee

Die Maximum-Likelihood-Methode wählt den Parameter, unter dem die beobachteten Daten die größte Likelihood besitzen.

Für die Exponentialrate erhalten wir:

$$
\hat\lambda_{\mathrm{MLE}}=\frac1{\hat\mu}.
$$

### 4a) MLE der Rate und des Mittelwerts

**Übersetzung der Aufgabe:**

Leite den MLE für $\lambda$ aus der Log-Likelihood her und überprüfe, dass ein Maximum vorliegt. Bestimme außerdem den MLE für den Mittelwert $\theta=1/\lambda$.

**Einfach erklärt:**

Wenn die beobachteten Wartezeiten im Durchschnitt kurz sind, schätzen wir eine hohe Rate. Wenn sie lang sind, schätzen wir eine niedrige Rate.

**Lösung:**

Setze

$$
T=\sum_{i=1}^{l}z_i.
$$

Die Log-Likelihood ist:

$$
\ell(\lambda)=l\log\lambda-\lambda T.
$$

Ableiten und gleich null setzen:

$$
\ell'(\lambda)=\frac l\lambda-T=0.
$$

Damit:

$$
\boxed{
\hat\lambda_{\mathrm{MLE}}
=
\frac lT
=
\frac1{\hat\mu}
}.
$$

Da

$$
\ell''(\lambda)=-\frac l{\lambda^2}<0,
$$

liegt für $T>0$ das eindeutige Maximum vor.

Weil $\theta=1/\lambda$ eine umkehrbare Transformation ist, gilt:

$$
\boxed{
\hat\theta_{\mathrm{MLE}}
=
\frac1{\hat\lambda_{\mathrm{MLE}}}
=
\hat\mu
}.
$$

**Bedeutung:** Der MLE der mittleren Wartezeit ist der beobachtete Durchschnitt; der MLE der Rate ist dessen Kehrwert.

### 4b) Verhalten bei endlichen Stichproben

**Übersetzung der Aufgabe:**

Simuliere für $\lambda=2$ jeweils 10 000 Stichproben der Größen $l=5,10,50,500$.

Vergleiche den MLE mit dem bias-korrigierten Schätzer

$$
\tilde\lambda
=
\frac{l-1}{l}\hat\lambda_{\mathrm{MLE}}.
$$

Berechne Bias, Varianz, MSE und $l\operatorname{Var}$. Vergleiche mit der CRLB und diskutiere Unverzerrtheit, Konsistenz und Effizienz.

**Einfach erklärt:**

Wir schätzen dieselbe Rate sehr oft mit neuen Stichproben und prüfen:

- Liegen die Schätzungen im Durchschnitt richtig?
- Wie stark schwanken sie?
- Wie groß ist ihr Gesamtfehler?

Die Korrektur multipliziert den MLE mit einem Faktor kleiner als 1, um seine Überschätzung auszugleichen.

**Theoretische Lösung für $l>2$:**

| Größe | MLE | Bias-korrigierter Schätzer |
|---|---|---|
| Schätzer | $l/T$ | $(l-1)/T$ |
| Bias | $\lambda/(l-1)$ | $0$ |
| Varianz | $\lambda^2l^2/((l-1)^2(l-2))$ | $\lambda^2/(l-2)$ |
| MSE | $\lambda^2(l+2)/((l-1)(l-2))$ | $\lambda^2/(l-2)$ |

Diese Formeln folgen aus $T\sim\operatorname{Gamma}(l,\lambda)$ und den inversen Momenten:

$$
\mathbb E[T^{-1}]=\frac{\lambda}{l-1},
\qquad
\mathbb E[T^{-2}]
=
\frac{\lambda^2}{(l-1)(l-2)}.
$$

**Interpretation:**

Der MLE überschätzt die Rate im Durchschnitt. Sein Bias ist bei kleinen Stichproben größer und geht für große Stichproben gegen null.

Der korrigierte Schätzer ist unverzerrt. Seine Varianz liegt aber über der CRLB:

$$
\frac{\lambda^2}{l-2}>\frac{\lambda^2}{l}.
$$

Er ist deshalb bei endlichen Stichproben nicht effizient. Auch der MLE ist bei endlichen Stichproben nicht effizient im Sinne eines unverzerrten Schätzers, der die CRLB erreicht.

Beide Schätzer sind:

- **konsistent:** Sie nähern sich mit wachsender Stichprobe dem wahren Parameter;
- **asymptotisch effizient:** Ihre Varianz nähert sich relativ zur CRLB dem optimalen Wert.

Insbesondere gilt für beide:

$$
l\operatorname{Var}(\hat\lambda)\longrightarrow\lambda^2=4.
$$

Hier verbessert die Bias-Korrektur sowohl Varianz als auch MSE. Das unterscheidet sich von Aufgabe 2: Dort erhöhte die Korrektur zur unverzerrten Varianzschätzung den MSE.

**Bedeutung:** Bias allein entscheidet nicht über die Qualität. Entscheidend ist

$$
\operatorname{MSE}=\operatorname{Varianz}+\operatorname{Bias}^2.
$$

```python
import numpy as np

rng = np.random.default_rng(0)
true_rate = 2.0
sample_sizes = [5, 10, 50, 500]
trials = 10_000

# Für Aufgabe 5c speichern.
sample_sums = {}

print("n | Estimator | Bias | Variance | MSE | CRLB | n*Variance")

for n in sample_sizes:
    z = rng.exponential(
        scale=1 / true_rate,
        size=(trials, n),
    )

    T = z.sum(axis=1)
    sample_sums[n] = T

    estimators = {
        "MLE": n / T,
        "Corrected": (n - 1) / T,
    }

    for name, estimates in estimators.items():
        bias = estimates.mean() - true_rate
        variance = estimates.var(ddof=0)
        mse = np.mean((estimates - true_rate)**2)
        crlb = true_rate**2 / n

        print(
            f"{n} | {name} | {bias:.6f} | "
            f"{variance:.6f} | {mse:.6f} | "
            f"{crlb:.6f} | {n * variance:.6f}"
        )

# Aufgabe 5: Bayesianische Schätzung mit einem Gamma-Prior

## Grundidee

Bayesianische Schätzung kombiniert Vorwissen mit beobachteten Daten:

$$
\text{Posterior}
\propto
\text{Likelihood}\times\text{Prior}.
$$

- **Prior:** Unser Vorwissen über den unbekannten Parameter.
- **Likelihood:** Wie gut verschiedene Parameterwerte zu den Daten passen.
- **Posterior:** Unsere aktualisierte Einschätzung nach Beobachtung der Daten.

Wir verwenden den Prior

$$
\lambda\sim\operatorname{Gamma}(\alpha,\beta),
$$

wobei $\beta$ der **Rateparameter** ist. Der Prior-Mittelwert beträgt $\alpha/\beta$.

## 5a) Posterior herleiten

### Übersetzung

Zeige, dass ein Gamma-Prior für die Exponentialrate nach Beobachtung der Daten wieder zu einer Gamma-Verteilung führt.

### Einfach erklärt

Die Gamma-Verteilung lässt sich hier leicht aktualisieren:

- Zum Shape-Parameter addieren wir die Anzahl der Beobachtungen.
- Zum Rateparameter addieren wir die Summe der beobachteten Wartezeiten.

### Lösung

Setze $T=\sum_{i=1}^{l}z_i$.

Die Likelihood ist, bis auf Faktoren unabhängig von $\lambda$,

$$
p(\mathbf Z\mid\lambda)
\propto
\lambda^l e^{-T\lambda}.
$$

Der Prior ist

$$
p(\lambda)
\propto
\lambda^{\alpha-1}e^{-\beta\lambda}.
$$

Durch Multiplikation erhalten wir

$$
p(\lambda\mid\mathbf Z)
\propto
\lambda^{\alpha+l-1}e^{-(\beta+T)\lambda}.
$$

Das ist die Form einer Gamma-Verteilung:

$$
\boxed{
\lambda\mid\mathbf Z
\sim
\operatorname{Gamma}(\alpha+l,\beta+T)
}.
$$

Der Prior heißt **konjugiert**, weil Prior und Posterior zur selben Verteilungsfamilie gehören.

## 5b) Posterior-Mittelwert und MAP

### Übersetzung

Berechne Posterior-Mittelwert und MAP-Schätzer und vergleiche sie mit dem MLE. Interpretiere die Priorparameter als Pseudo-Beobachtungen. Zeige, was bei großen Stichproben passiert und wann der MAP immer dem MLE entspricht.

### Einfach erklärt

Der Posterior liefert eine ganze Verteilung für die mögliche Rate. Daraus wählen wir einen einzelnen Schätzwert:

- **Posterior-Mittelwert:** den Durchschnitt unter dem Posterior.
- **MAP:** den Wert mit der höchsten Posterior-Dichte.

### Lösung

Aus der Gamma-Verteilung folgen:

$$
\boxed{
\mathbb E[\lambda\mid\mathbf Z]
=
\frac{\alpha+l}{\beta+T}
}
$$

und für $l\geq1$:

$$
\boxed{
\hat\lambda_{\mathrm{MAP}}
=
\frac{\alpha+l-1}{\beta+T}
}.
$$

Zum Vergleich lautet der MLE:

$$
\hat\lambda_{\mathrm{MLE}}=\frac lT.
$$

Der Posterior-Mittelwert lässt sich schreiben als

$$
\mathbb E[\lambda\mid\mathbf Z]
=
\frac{T}{T+\beta}\hat\lambda_{\mathrm{MLE}}
+
\frac{\beta}{T+\beta}\frac{\alpha}{\beta}.
$$

Er ist also ein gewichteter Durchschnitt aus MLE und Prior-Mittelwert.

### Pseudo-Beobachtungen

Die Log-Likelihood enthält

$$
l\log\lambda-T\lambda.
$$

Der Log-Prior ergänzt

$$
(\alpha-1)\log\lambda-\beta\lambda.
$$

Für den MAP wirkt der Prior daher wie ein zusätzlicher Beitrag von $\alpha-1$ Beobachtungen mit gesamter Wartezeit $\beta$.

Die Interpretation als positive Anzahl ist für $\alpha>1$ sinnvoll. Nichtganzzahlige Werte entsprechen dabei einer rechnerischen Analogie.

### Verhalten bei großen Stichproben

Mit $T=l\hat\mu$ gilt:

$$
\mathbb E[\lambda\mid\mathbf Z]
=
\frac{1+\alpha/l}{\hat\mu+\beta/l},
$$

$$
\hat\lambda_{\mathrm{MAP}}
=
\frac{1+(\alpha-1)/l}{\hat\mu+\beta/l}.
$$

Für feste Priorparameter und einen festen positiven Stichprobenmittelwert folgt:

$$
\boxed{
\mathbb E[\lambda\mid\mathbf Z],
\hat\lambda_{\mathrm{MAP}}
\longrightarrow
\frac1{\hat\mu}
=
\hat\lambda_{\mathrm{MLE}}
}.
$$

Der Einfluss des Priors nimmt also mit wachsender Stichprobe ab.

### Wann entspricht der MAP immer dem MLE?

Die Gleichheit für alle Stichproben erfordert

$$
\boxed{\alpha=1,\qquad\beta=0}.
$$

Das entspricht dem formalen Prior

$$
p(\lambda)\propto1,\qquad\lambda>0.
$$

Dieser Prior ist **improper**: Er kann auf dem gesamten positiven Bereich nicht auf 1 normiert werden und ist deshalb keine reguläre Gamma-Verteilung.

## 5c) Passender und unpassender Prior

### Übersetzung

Vergleiche bei der wahren Rate $\lambda=2$ den MLE mit dem Posterior-Mittelwert für:

- einen passenden Prior $\operatorname{Gamma}(4,2)$ mit Mittelwert $2$;
- einen unpassenden Prior $\operatorname{Gamma}(20,2)$ mit Mittelwert $10$.

Verwende jeweils 10 000 unabhängige Stichproben mit $l=5,10,50,500$ Beobachtungen und vergleiche Bias, Varianz und MSE.

### Einfach erklärt

Der passende Prior zieht die Schätzung zum richtigen Wert. Der falsche Prior zieht sie zu einem zu hohen Wert.

Bei wenigen Daten ist dieser Einfluss groß. Bei vielen Daten wird er kleiner.

### Simulationscode

Der folgende Code ist unabhängig ausführbar.

```python
import numpy as np

rng = np.random.default_rng(0)

true_rate = 2.0
sample_sizes = [5, 10, 50, 500]
trials = 10_000

priors = {
    "Correct prior": (4, 2),
    "Wrong prior": (20, 2),
}

print("n | Estimator | Bias | Variance | Bias^2 | MSE")

for n in sample_sizes:
    z = rng.exponential(
        scale=1 / true_rate,
        size=(trials, n),
    )
    T = z.sum(axis=1)

    estimators = {"MLE": n / T}

    for name, (alpha, beta) in priors.items():
        estimators[name] = (alpha + n) / (beta + T)

    for name, estimates in estimators.items():
        bias = estimates.mean() - true_rate
        variance = estimates.var(ddof=0)
        mse = np.mean((estimates - true_rate)**2)

        print(
            f"{n} | {name} | {bias:.6f} | "
            f"{variance:.6f} | {bias**2:.6f} | {mse:.6f}"
        )
```

Bei $l=5$ sinkt der MSE von ungefähr $2.3285$ beim MLE auf $0.2676$. Der passende Prior reduziert sowohl Bias als auch Varianz deutlich.

Er macht den Schätzer jedoch nicht exakt unverzerrt: Der Bias beträgt noch ungefähr $0.1189$.

**Unpassender Prior:**

Bei $l=5$ beträgt der Bias ungefähr $3.8859$. Sein Quadrat ist ungefähr $15.1001$ und dominiert den MSE von $17.0556$.

Der falsche Prior reduziert die Varianz nicht durchgehend. Bei $l=10,50,500$ liegt sie sogar über der Varianz des MLE.

**Große Stichprobe:**

Bei $l=500$ sind die MSEs von MLE und passendem Prior fast gleich:

- MLE: ungefähr $0.008146$;
- passender Prior: ungefähr $0.008015$.

Der falsche Prior hat mit ungefähr $0.013284$ weiterhin einen höheren MSE. Sein Einfluss ist aber deutlich kleiner als bei $l=5$.

### Erklärung mit der Bias-Varianz-Zerlegung

Es gilt:

$$
\boxed{
\operatorname{MSE}
=
\operatorname{Varianz}
+
\operatorname{Bias}^2
}.
$$

Ein passender Prior kann den Gesamtfehler reduzieren. Ein falscher Prior kann dagegen einen großen Bias erzeugen und dadurch den Gesamtfehler erhöhen.

## Zusammenfassung

Bayesianische Schätzung verbindet Vorwissen mit Daten. Ein passender Prior hilft besonders bei kleinen Stichproben. Ein unpassender Prior kann die Schätzung stark verschlechtern. Mit wachsender Stichprobengröße nimmt der Einfluss eines festen Priors ab.
